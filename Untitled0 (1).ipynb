```python
# Graph representation using dictionary
graph = {
    '5': ['3', '7'],
    '3': ['2', '4'],
    '7': ['8'],
    '2': [],
    '4': ['8'],
    '8': []
}

# List to keep track of visited nodes
visited = []

# Queue for BFS
queue = []

# BFS function
def bfs(visited, graph, node):
    visited.append(node)
    queue.append(node)

    while queue:  # Loop until queue is empty
        m = queue.pop(0)
        print(m, end=" ")

        for neighbour in graph[m]:
            if neighbour not in visited:
                visited.append(neighbour)
                queue.append(neighbour)

# Driver code
print("Following is the Breadth-First Search:")
bfs(visited, graph, '5')
```

```python
# Function to print the parent of each node (DFS)
def printParents(node, adj, parent):
    # Root node
    if parent == 0:
        print(node, "-> Root")
    else:
        print(node, "->", parent)

    # Traverse children
    for cur in adj[node]:
        if cur != parent:
            printParents(cur, adj, node)


# Function to print the children of each node (BFS)
def printChildren(Root, adj):
    q = []
    q.append(Root)

    vis = [0] * len(adj)

    while len(q) > 0:
        node = q.pop(0)
        vis[node] = 1

        print(node, "-> ", end=" ")

        for cur in adj[node]:
            if vis[cur] == 0:
                print(cur, end=" ")
                q.append(cur)

        print()


# Function to print leaf nodes
def printLeafNodes(Root, adj):
    for i in range(1, len(adj)):
        if len(adj[i]) == 1 and i != Root:
            print(i, end=" ")
    print()


# Function to print degree of each node
def printDegrees(Root, adj):
    for i in range(1, len(adj)):
        print(i, ": ", end="")

        if i == Root:
            print(len(adj[i]))
        else:
            print(len(adj[i]) - 1)


# Driver Code
N = 7
Root = 1

# Create adjacency list
adj = []
for i in range(N + 1):
    adj.append([])

# Creating the tree
adj[1].append(2)
adj[2].append(1)

adj[1].append(3)
adj[3].append(1)

adj[1].append(4)
adj[4].append(1)

adj[2].append(5)
adj[5].append(2)

adj[2].append(6)
adj[6].append(2)

adj[4].append(7)
adj[7].append(4)

# Outputs
print("The parents of each node are:")
printParents(Root, adj, 0)

print("\nThe children of each node are:")
printChildren(Root, adj)

print("\nThe leaf nodes of the tree are:")
printLeafNodes(Root, adj)

print("\nThe degrees of each node are:")
printDegrees(Root, adj)
```

```python
from collections import deque

# Weighted Graph (vertex : [(neighbor, cost)])
graph = {
    'A': [('B', 2), ('C', 4)],
    'B': [('A', 2), ('D', 7), ('E', 3)],
    'C': [('A', 4), ('F', 5)],
    'D': [('B', 7)],
    'E': [('B', 3), ('G', 6)],
    'F': [('C', 5)],
    'G': [('E', 6)]
}

# BFS with cost calculation
def BFS_with_cost(start, goal):
    visited = set()
    q = deque()

    # (current_node, path, total_cost)
    q.append((start, [start], 0))

    while q:
        node, path, cost = q.popleft()

        # If goal reached
        if node == goal:
            print("Start:", start, "Goal:", goal)
            print("Path:", " -> ".join(path))
            print("Total Cost:", cost)
            print("---------------------------")
            return

        if node not in visited:
            visited.add(node)

            for neighbor, weight in graph[node]:
                if neighbor not in visited:
                    q.append((neighbor, path + [neighbor], cost + weight))

    print("No path found from", start, "to", goal)
    print("---------------------------")


# Driver Code (Different variations of start and goal)

BFS_with_cost('A', 'G')
BFS_with_cost('A', 'F')
BFS_with_cost('B', 'G')
BFS_with_cost('C', 'D')
```
